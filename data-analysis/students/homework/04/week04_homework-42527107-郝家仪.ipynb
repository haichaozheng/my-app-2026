# 第三部分：数据采集

```python
import tushare as ts
# 为了批改方便，我这里将tushare token直接放到代码中
token = "5c426f5af978fd42fe0f81fe7e60c02ada33d6c79347b61ccb44f521"
ts.set_token(token=token)
pro = ts.pro_api()
```

```python

# 1. 获取财务数据（利润表 - 2019到2023年）
print("正在获取财务数据...")
income_data = pro.income(
    ts_code='600519.SH', 
    start_date='20190101', 
    end_date='20231231',
    fields='ts_code,end_date,revenue'
)

# 2. 获取行情数据（日线 - 2019到2023年）
print("正在获取行情数据...")
price_data = pro.daily(
    ts_code='600519.SH', 
    start_date='20190101', 
    end_date='20231231',
    fields='ts_code,trade_date,close'
)

print("数据采集完成！")
# 展示前5行数据
display(income_data.head())
display(price_data.head())
```

# 第四部分：数据分析过程

```python
import pandas as pd
import matplotlib.pyplot as plt
# 1. 处理财务数据：只取每年的年报数据（日期以1231结尾），并转为“亿元”
income_data['year'] = income_data['end_date'].str[:4]
annual_income = income_data[income_data['end_date'].str.endswith('1231')].copy()
# 去除重复数据，并取营收（为了防止有重复数据，只保留一条）
annual_income = annual_income.groupby('year')['revenue'].max().reset_index()
# 将单位转换为“亿元”方便观看
annual_income['revenue_yi'] = annual_income['revenue'] / 100000000

# 2. 处理行情数据：取每年最后一天的收盘价
price_data['year'] = price_data['trade_date'].str[:4]
annual_price = price_data.groupby('year')['close'].last().reset_index()
annual_price.rename(columns={'close': 'year_end_price'}, inplace=True)

# 3. 合并两个数据表
df = pd.merge(annual_income, annual_price, on='year')
print("合并后的数据如下：")
display(df)

# 4. 画图（双轴图：柱状图表示营收，折线图表示股价）
fig, ax1 = plt.subplots(figsize=(10, 6))

# 画营收柱状图
ax1.bar(df['year'], df['revenue_yi'], color='skyblue', alpha=0.7, label='年营业收入(亿元)')
ax1.set_xlabel('年份', fontsize=12)
ax1.set_ylabel('营业收入 (亿元)', color='blue', fontsize=12)
ax1.tick_params(axis='y', labelcolor='blue')

# 创建双轴，画股价折线图
ax2 = ax1.twinx()
ax2.plot(df['year'], df['year_end_price'], color='red', marker='o', linewidth=2, label='年末收盘价(元)')
ax2.set_ylabel('年末收盘价 (元)', color='red', fontsize=12)
ax2.tick_params(axis='y', labelcolor='red')

plt.title('贵州茅台2019-2023年营业收入与年末股价走势对比', fontsize=14)
fig.tight_layout()
plt.show()

# 5. 计算相关系数
correlation = df['revenue_yi'].corr(df['year_end_price'])
print(f"营业收入与年末股价的相关系数为: {correlation:.2f}")
```
